# Features Scalling 

    Types of Features scalling :
    1. Normalization (L2 Norm)
    2. Standard Scaling 

```python
from sklearn.preprocessing import Normalizer 
import pandas as pd 
```

```python
d={
    "Age":[25,30,35,40],
    "Salary":[30000,40000,50000,60000]
}
df=pd.DataFrame(d)
df
```

# Normalizer 

```python
ns=Normalizer()
v1=ns.fit_transform(df)
df1=pd.DataFrame(v1,columns=["Scaled_age","Scaled_salary"])
df1
```

# Standard Scaling 

```python
from sklearn.preprocessing import StandardScaler 
```

```python
sc=StandardScaler()
v2=sc.fit_transform(df)
df2=pd.DataFrame(v2,columns=["Scaled_age","Scaled_salary"])

pd.concat([df1,df2],axis=1)
```
