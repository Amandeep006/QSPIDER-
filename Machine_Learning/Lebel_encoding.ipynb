# Label Enconding 

```python
from sklearn.preprocessing import LabelEncoder as le
from sklearn.preprocessing import OneHotEncoder as on
import pandas as pd
```

```python
d={
    "Climate":["Sunny","Rainy","Windy","Cloudy","Rainy","Stormy"]
}
df=pd.DataFrame(d)
df
```

# LabelEncoder 

```python
# Here we create a object and also label is the reference variable.
label=le()
df["encoded_climate"]=label.fit_transform(df["Climate"]) # fit_transform() is used to transform a string into ASCII Value 
df
```

# OneHot Encoder 

```python
onehot=on()
Encoded_value=onehot.fit_transform(df[["Climate"]])
```

```python
# Sparse Matrix : There is no 1 element present in the matrix.
v1=Encoded_value.toarray()
v1
```

```python
# It is used to find the name of the columns according to the given dataframe.
cl= onehot.get_feature_names_out()
cl
```

```python
df1=pd.DataFrame(v1,columns=cl)
df1
```

```python
df1
```

# Ordinal Encoder 

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
d1={
    "Salary":["High","Low","Medium","Low","High"]
}
df2=pd.DataFrame(d1)
df2
```

```python
code=OrdinalEncoder(categories=[["Low","Medium","High"]])
df2["Encoded_salary"]=code.fit_transform(df2[["Salary"]])
df2
```
