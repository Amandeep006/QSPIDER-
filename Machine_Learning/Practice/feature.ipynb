```python
from sklearn.preprocessing import LabelEncoder 
import pandas as pd 
```

```python
df=pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\Qspider\datasets\Data-Analysis-Datasets\Datasets-Practice\titanic.csv")
df
```

```python
label=LabelEncoder()
df["Encoded_name"]=label.fit_transform(df["name"])
df
```

```python
# Surname of the name column 

surname=df["name"].str.split(",").str[0]
sur=pd.DataFrame(surname)

sur["label_encoding"]=label.fit_transform(sur["name"])

sur["name"]
# surname
```

```python
# Title of name column 

title=df["name"].str.split(",").str[1].str.split(".").str[0]
tit=pd.DataFrame(title,columns=["title"])

tit["label_encoding"]=label.fit_transform(tit["title"])
tit
```
