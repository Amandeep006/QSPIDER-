```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

# Plot()
    It is used to create 2D graph.
    Syntax : 
               plt.plot(x,y) 

```python
x=np.arange(1,11)
y=x**2
z=y//x

print(x)
print(y)
print(z)
```

```python
# plt.figure(figsize=(w,h)):- 
                #            It is used to adjust the height and widht of chart.

plt.figure(figsize=(15,5))
plt.plot(x,y,label="y=x**2",marker="o",markersize=7,color="Purple",markerfacecolor="Black",markeredgecolor="Yellow",linewidth=5,linestyle="--")
plt.plot(x,z,label="z=y/x",marker="o",markersize=7,color="Red",markerfacecolor="Black",markeredgecolor="Yellow",linewidth=5,linestyle="dotted")
plt.legend(title="values",fontsize=15,loc="center")# IF legend is present then only the label going to work.
plt.xlabel("values of x", fontsize=22)
plt.ylabel("values of y",fontsize=22)
plt.grid()
plt.title("Graph", fontsize=20,color="Pink")


# Annotate
for i in range(len(x)):
    plt.annotate(f"({x[i]},{y[i]})",(x[i],y[i]),va="top",ha="left") 
    # va - Vertical alignment  & ha - horizontal alignment

for i in range(len(x)):
    plt.annotate(f"({x[i]},{z[i]})",(x[i],z[i]),va="top",ha="left")
```

# Subplot

Syntax : 
    
    plt.subplot(no_of_rows,no_of_cols,position)

```python
plt.subplot(3,3,1)
plt.subplot(3,3,2)
plt.subplot(3,3,3)
plt.subplot(3,3,4)
plt.plot(x,y)
plt.xlabel("x-label")
for  i in range(5,10):
    plt.subplot(3,3,i)
plt.tight_layout() # It  will create sufficent gap between all the graphs.
```

# Pie Chart
    
    To represent the data in the form of percentage.


Syntax:  
   
    plt.pie(data,label=[],
            autopct="%0.1f%%",
            color=[],
            shadow=True,
            explode())


```python
viewers=[1500,2000,1700,500,1000]
lab=["comedy","Action","Horror","ROmance","Drama"]
plt.pie(viewers,labels=lab
               ,autopct="%0.5f%%"
               ,colors=["Red","Yellow","Blue","Purple","Orange"]
               ,shadow=True
               ,explode=[0.8,0.1,0.2,0.4,0.3])
```
