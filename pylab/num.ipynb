1. 
LIST : It is collaction of homogenous and heterogenous data inclosed between square braces [].

2. 
LIST : We can perform limited operation on list.
ARRAY : We can perform almost all operations on array.

3. 
LIST : It stores the data in discontinous memory allocation. 
ARRAY : It stores the data in continous memory allocation.

4. 
LIST : Acessing of values is slow in list.
ARRAY : Accessing of values is fast in list.

```python
# Boolean Filtering 
# Syntax : var[condition]
 

ages=np.array([[35,22,75,45,15,26,60,33,51,67],[33,62,45,11,36,27,33,22,78,44]])
ages
```

```python
# Filter out those ages which are greater than 60.
ages[ages>60]
```

```python
# Filter out the  ages which are less than 18
ages[ages<18]
```

```python
# Filter out the ages between 25 to 50.
ages[(ages>25)&(ages<50)]
```

```python
# Filter out the ages which are less than 18 and more than 60.
ages[~((ages<18) | (ages>60))]
```

```python
# Filter out the ages which are in between 18 to 60 
ages[(ages>18)&(ages<60)]
```
