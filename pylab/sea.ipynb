#  Difference between Matplotlib and Seaborn 

# matplot : 
1. It is used to plot simple data.
2. It has limited Plots.
3. It has attributes.

# Seaborn 
1. It is used to plot complex data.
2. It has more number of Plots.
3. It has more number of attributes.

# Types of Plots in Seaborn .

    1. Distribution Plots
    2. Chatogorical Plots 
    3. Matrix Plots 
    4. Regression Plots

# Types of Plots based on number of columns used in plot.

    1. Uni-variate Plot : It allows us to plot the graph based on one column.
    2. Bi-variate Plot :  It allows us to plot the graph based on two column.
    3. Multi-variate Plot : It allows us to plot the graph based on multiple column.

# histplot(): 

    It is used  to represent count the unique values.

Syntax :

    sns.histplot(x="col_name",data-var,color="any",hue="category_col)

```python
import seaborn as sns 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ti=pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\Qspider\datasets\Data-Analysis-Datasets\Datasets-Practice\tips.csv")
```

```python
sns.histplot(x="sex",data=ti,color="orange")
```

```python
# palette-  inferno,viridis, magma, Set1,Set2,Set3,coolwarm, Reds, Green, Purples
a=sns.histplot(x="sex",data=ti,palette="Set1",hue="smoker")
```

```python
# We can use loop for showing the bar levels
a=sns.histplot(x="sex",data=ti,palette="viridis",hue="smoker")
for i in a.containers:
    a.bar_label(i)
```

```python
qu=pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\Qspider\datasets\Data-Analysis-Datasets\Datasets-Practice\winequalityN.csv")

sns.histplot(x="quality",data=qu,palette="Reds",weights=7,hue="type",bins=6)
plt.show()
```

```python
hr=pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\Qspider\datasets\Data-Analysis-Datasets\Datasets-Practice\hr_data.csv")
a=sns.histplot(x="salary",data=hr,hue="salary")
for i in a.containers:
    a.bar_label(i,label_type="center",size=15)
```

# Jointplot(): 

    It is the combination of scatter and list. It is used to identify the reltionship between two columns.

Syntax : 

    sns.jointplot(x="col_name,y="col_name,data=var)

```python
sns.jointplot(x="total_bill",y="tip",data=ti,color="Orange",hue="sex")
plt.show()
# kde=kernal density estimate  

```

```python
win=pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\Qspider\datasets\Data-Analysis-Datasets\Datasets-Practice\winequalityN.csv")
sns.jointplot(x="pH",y="density",data=win,palette="viridis",hue="quality")
```

# Categorical Plot :- 

    It is used to analyse the data based on categorical columns.

Types of Categorical columns : 
1. Count plot 
2. Bar plot

4. Scatter plot 
5. Violin plot

# 1. Countplot ():-

    It is used ot count the unique values.

Syntax : 

    sns.countplot(x="col_name",data=var)


```python
ti["smoker"].value_counts()

c=sns.countplot(x="smoker",data=ti,hue="sex")
for i in c.containers:
    c.bar_label(i)
```

```python
t=pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\Qspider\datasets\Data-Analysis-Datasets\Datasets-Practice\titanic.csv")
b=sns.countplot(x="sex",data=t)
for i in b.containers:
    b.bar_label(i)
```

```python
d=sns.countplot(x="type",data=win)
for i in d.containers:
    d.bar_label(i)
```

# 2. barplot():

    It is used to know  the average/mean value.

Syntax :

    sns.barplot(x="col_name",y="col_name",data=var)

```python
e=sns.barplot(x="type",y="quality",data=win,hue="quality",errorbar=None,palette="viridis")
for i in e.containers:
    e.bar_label(i)
```

```python
f=sns.barplot(x="sex",y="age",data=t,hue="survived")
for i in f.containers:
    f.bar_label(i)
```
