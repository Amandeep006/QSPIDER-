```python
#  145.Wap to check whether the number is spy or not.
#  Ex: 123→1+2+3=1*2*3

def spy(n):
    sum=0
    mul=1
    while n>0:
        rem=n%10
        sum+=rem
        mul*=rem
        n//=10

    if sum==mul:
        print("The given number is spy")
    else:
        print("THe given number is not spy.")



n=int(input("Enter any number : "))
spy(n)
```

```python
#  144.Wap to check whether the number is xylem or not.
#  Ex: 1234→1+4=2+3

def xylem(n):
    l=[]
    while n>0:
        rem=n%10
        l.append(rem)
        n//=10
    sum=l[0]+l[-1]
    r=0
    for i in l:
        if i==l[0] or i==l[-1]:
            continue
        else:
            r+=i

    if r==sum:
        print("The given number is Xylem")
    else:
        print("The given number is not Xylem.")


n=int(input("Enter any number : "))
xylem(n)
```
