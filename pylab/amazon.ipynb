# Data cleaning operations 

```python
# see the information 
ama.info()
ama.head()
```

```python
# To rename a=the column name :
var={
    "ano":"year",
    "mes":"months",
    "estado":"state",
    "numero":"no_of_fires",
    "encontro":"date"
}

ama.rename(columns=var,inplace=True)
```

```python
# TO replace the specific value form the dataset ;
ma={
    'Janeiro':"January",
    'Fevereiro':"February",
    'Março':"March",
    'Abril':"April",
    'Maio':"May",
    'Junho':"June",
    'Julho':"July",
    'Agosto':"August",
    'Setembro':"September",
    'Outubro':"October",
    'Novembro':"November",
    'Dezembro':"December"
}

ama["months"]=ama["months"].map(ma)

ama["months"].unique()
```

```python
# Find how many null values are present.
ama["no_of_fires"].isnull().sum()
```

```python
# find how many duplicates column in a dataset.
# ama.duplicated().sum()

ama.drop_duplicates(inplace=True)
```

```python
# To remove the fires from the no. of fires column 
ama["no_of_fires"]=ama["no_of_fires"].str.strip("Fires")

```

```python
# replace  the data of no_of_fires from string to int
ama["no_of_fires"]=ama["no_of_fires"].astype(float)
```

```python
# To fill null values
med=round(ama["no_of_fires"].median())


ama["no_of_fires"]=ama["no_of_fires"].fillna(med)
```

```python
ama["no_of_fires"]=ama["no_of_fires"].astype(int)
```
