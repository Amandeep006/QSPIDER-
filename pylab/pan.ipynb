# Loc and iloc
 loc--> Label based indexing 
 iloc--> Location based indexing 

loc --> Label based indexing  
  syntax :  
        var.loc[index] -- > to access single value 
        var.loc[[index1,index2---indexN]]--> to access multiple values

iloc ---> Location based indexing 
  syntax :
        var.iloc[index] --> to access single value 
        var.iloc[[index1,index2,index3---indexN]] ---> To access multiple values 

```python
n=["app","man","gra","oran"]
a=pd.Series(n)
s=pd.Series(n,index=["a","b","c","d"])

# Loc method 
a.loc[[0,1,3]]
s.loc[["a","c","b","d"]]
a.loc[[1,3]]
s.loc[::-1]
s.loc["a"::2]
a.loc[0::2]
a.loc[::-1]

# iloc method 
a.iloc[0]
a.iloc[2]
a.iloc[[0,1,3,2]]
s.iloc[[3,1,0,2]]
```

```python
# Indexing and Slicing on Data
# loc and iloc 

# loc --> label based indexing 
  # 1. var.loc[index] or var.loc[index,"col_name"]
  # 2. var.loc[[index1,index2---indexN]] or var.loc[[index1,index2---indexN],[col1,col2---colN]]
  # 3. var.loc[sv:ev:up]
  # 4. var.loc[sv:ev:up,"col_name"] or var.loc[sv:ev:up,[col1,col2---colN]]


tips.loc[1]
tips.loc[[1,2,4,5]]
tips.loc[1::3]
tips.loc[::-1,["tip","smoker","day"]]

#iloc ---> Location based indexing 
   # 1. var.iloc[index]
   # 2. var.iloc[[index1, index2---indexN]] or var.iloc[[index1,index2---indexN],[col_index1,col_index2----col_indexN]]
   # 3. var.iloc[sv:ev:up]
   # 4. var.iloc[sv:ev:up,[col_index1, col_index2---col_indexN]]
   # 5. var.iloc[sv:ev:up,sv:ev:up]
                # row  slice, calumn slice

tips.iloc[1]
tips.iloc[[1,2,5,9],[1,3,4]] # we are not giving here column name , if you want to take column just give column index value.
tips.iloc[0:8,[1,2,3]] 
```
