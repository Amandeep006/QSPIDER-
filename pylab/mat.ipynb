# Data Visualization

    It is process of coverting the data into Pictorial or graphical format to understand data in better manner.

There are three ways to visualize the data.

    1. Matplotlib (library)
    2. Seaborn (library)
    3. Power B.I. (Tool)

```python
import matplotlib.pyplot as plt 
import numpy as np
import pandas as pd
```

# Plot()

    It is used to plot 2D graph.

```python
# Syntax : plt.plot(var1,var2,marker="o,*,s,h,p",color="Red",linestyle="dashed/dotted/solid,linewidth=int,markersize=int)

x=np.arange(1,11)
y=x**2
z=x//y

print(x)
plt.figure(facecolor="cyan") # figsize=(width,height)
plt.plot(x,y,marker="o",color="Red",markerfacecolor="Yellow",markeredgecolor="Black",linewidth=3,linestyle="--",markersize=10)
plt.grid()
plt.xlabel("X-values",fontsize=12,color="Black")
plt.ylabel("Y-label",fontsize=12,color="Black")
plt.title("Graph",fontsize=18,color="Red")
plt.gca().set_facecolor("White") # Git current access


# Coordinate value of Graph :
for i in range(len(x)):
    plt.annotate(f"({x[i]},{y[i]})",(x[i],y[i]))
    #                 what to print, where to print
```

```python
x=np.arange(1,11)
y=x**2
z=x**3

print(x)
plt.figure(facecolor="cyan") # figsize=(width,height)
plt.plot(x,z,marker="h",color="Red",markerfacecolor="Yellow",markeredgecolor="Black",linewidth=3,linestyle="--",markersize=10)
plt.grid()
plt.xlabel("X-values",fontsize=12,color="Black")
plt.ylabel("Z-label",fontsize=12,color="Black")
plt.title("Graph",fontsize=18,color="Red")
plt.gca().set_facecolor("White") # Git current access


# Coordinate value of Graph :
for i in range(len(x)):
    plt.annotate(f"({x[i]},{z[i]})",(x[i],z[i]),fontsize=10,ha="right",va="bottom")
    #                 what to print, where to print
```
