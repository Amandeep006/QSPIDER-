```python
import matplotlib.pyplot as plt 
import numpy as np
```

# Plot()

    It is used to plot 2D graph.

```python
# Syntax : plt.plot(var1,var2,marker="o,*,s,h,p",color="Red",linestyle="dashed/dotted/solid,linewidth=int,markersize=int)

x=np.arange(1,11)
y=x**2
z=x//y

print(x)
plt.figure(figsize=(15,5))
plt.plot(x,y,marker="s",color="Red",markerfacecolor="Yellow",markeredgecolor="Black",linewidth=3,linestyle="--",markersize=10)
```
