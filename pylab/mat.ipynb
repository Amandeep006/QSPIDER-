# Data Visualization

    It is process of coverting the data into Pictorial or graphical format to understand data in better manner.

There are three ways to visualize the data.

    1. Matplotlib (library)
    2. Seaborn (library)
    3. Power B.I. (Tool)

```python
import matplotlib.pyplot as plt 
import numpy as np
import pandas as pd
```

# Plot()

    It is used to plot 2D graph.

```python
# Syntax : plt.plot(var1,var2,marker="o,*,s,h,p",color="Red",linestyle="dashed/dotted/solid,linewidth=int,markersize=int)
#Note :- Inside the label argument only works when we call the lengend() function otherwise it is wasted.

x=np.arange(1,11)
y=x**2
z=x//y

print(x)
plt.figure(facecolor="cyan") # figsize=(width,height)
plt.plot(x,y,marker="o",color="Red",markerfacecolor="Yellow",markeredgecolor="Black",linewidth=3,linestyle="--",markersize=10,label="Square-Values")
plt.plot(x,z,marker="s",color="Black",markerfacecolor="Yellow",markeredgecolor="Red",linewidth=3,linestyle="--",markersize=10,label="Cube-Values")
plt.grid()
plt.xlabel("X-values",fontsize=12,color="Black")
plt.ylabel("Y-label",fontsize=12,color="Black")
plt.title("Graph",fontsize=18,color="Red")
plt.legend(title="Co-ordinate")
plt.gca().set_facecolor("White") # Git current access


# Coordinate value of Graph :
for i in range(len(x)):
    plt.annotate(f"({x[i]},{y[i]})",(x[i],y[i]))
    #                 what to print, where to print
```

```python
x=np.arange(1,11)
y=x**2
z=x**3

print(x)
plt.figure(facecolor="cyan") # figsize=(width,height)
plt.plot(x,z,marker="h",color="Red",markerfacecolor="Yellow",markeredgecolor="Black",linewidth=3,linestyle="--",markersize=10)
plt.grid()
plt.xlabel("X-values",fontsize=12,color="Black")
plt.ylabel("Z-label",fontsize=12,color="Black")
plt.title("Graph",fontsize=18,color="Red")
plt.gca().set_facecolor("White") # Git current access


# Coordinate value of Graph :
for i in range(len(x)):
    plt.annotate(f"({x[i]},{z[i]})",(x[i],z[i]),fontsize=10,ha="right",va="bottom")
    #                 what to print, where to print
```

# Pie chart():
 
        It represents the data into percentage format.

Syntax : 
 
        plt.pie(data, labels=[],autopct="0.2f%%,color=[],shadow=True,
                     explode=())

```python
movie_cat=["comedy","action","horror","drama","romantic"]
viewers=[2000,1500,1000,2500,500]

plt.pie(viewers,labels=movie_cat,autopct="%0.0f%%",colors=["Red","Green","Yellow","Grey","cyan","orange"],shadow=True,explode=[0.1,0.2,0.3,0.2,0.1])
```

```python
ti=pd.read_csv(r"C:\Users\HP\OneDrive\Desktop\Qspider\datasets\Data-Analysis-Datasets\Datasets-Practice\titanic.csv")
sur=ti[ti["survived"]==0]["survived"].count()
nsur=ti[ti["survived"]==1]["survived"].count()
```

```python
s=[sur,nsur]
l=["Survived","Unsurvived"]

s=ti["survived"].value_counts()
plt.pie(s,labels=l,autopct="%0.1f%%",colors=["Red","Green"],shadow=True,explode=[0.1,0])
```
