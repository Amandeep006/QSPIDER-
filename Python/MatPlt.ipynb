MATPLOTLIB FOR VISUALIZATION 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

BAR-PLOT USING MATPLOTLIB 

```python
y=[98,68,96,61,88]
x=["part1","part2","part3","part4","part5"]
color=["red","blue","green","yellow","black"]

# plt.bar(x,y)

plt.bar(x,y,color=color)
plt.xlabel("Parts of the Harry Potter",fontsize=5)
plt.ylabel("Popularity",fontsize=5)
plt.title("Popularity of different parts of Harry Potter",fontsize=20)
```

LINE -PLOT MATPLOTLIB 

```python
y=[98,68,96,61,88]
x=["part1","part2","part3","part4","part5"]
y1=[500,300,400,250,350]
plt.plot(x,y,marker="o",label="week1",alpha=0.4)
plt.xlabel("Parts of the Harry Potter Series", fontsize=10)
plt.ylabel("Popularity",fontsize=10)
plt.title("Popularity of different parts of Harry Potter",fontsize=15)
plt.plot(x,y1,marker="o",color="red",label="week2",alpha=0.5)
plt.legend()
plt.show()

```

SCATTER PLOT MATPLOTLIB 

```python
x=np.random.randint(1,100,50)
y=np.random.randint(10,100,50)
color=np.random.randint(10,100,50)
plt.scatter(x,y,marker="o",c=color,cmap="cool")
plt.colorbar()
plt.show()
```

PIE-CHART MATPLOTLIB 

```python
brands=["apple","oneplus","google-pixcel","nokia","redmi"]
x=[22,35,30,31,10]
color=["red","orange","violet","green","pink"]
plt.pie(x,labels=brands,colors=color, shadow=True,startangle=180, autopct='%1.1f%%'  )
plt.show()
```

BOX-PLOT MATPLOTLIB

```python
# Sample data: Exam scores for two different classes
class_a = [55, 62, 67, 71, 74, 78, 82, 85, 89, 92, 95]
class_b = [60, 72, 74, 75, 78, 80, 81, 84, 85, 88, 100]

# Create the box plot
plt.boxplot([class_a, class_b], labels=['Class A', 'Class B'])
# Add labels and title
plt.ylabel('Exam Scores')
plt.title('Exam Score Distribution by Class')

plt.show()
```

HISTOGRAM MATPLOTLIB 

```python
x=np.random.randint(1,100,15)
print(x)
plt.hist(x,bins=len(x),edgecolor="black",color="pink")
plt.show()
```

VIOLIN PLOT MATPLOTLIB 

```python
a=[20,60,30,40,45,35,50,12,16]
plt.violinplot(a,showmedians=True)
plt.show()
```

STEM-PLOT MATPLOTLIB

```python
a=np.random.randint(1,100,15)
print(a)
plt.stem(a,linefmt="--",markerfmt="d",bottom=10,orientation="vertical") # or orientation ="horizontal"
plt.show()
```

STACKPLOT MATPLOTLIB

```python
days=[1,2,3,4,5,6,7]
nop1=[5,10,30,20,35,60,80]
nop2=[10,20,30,25,40,50,90]
nop3=[8,30,50,60,70,90,100]
plt.stackplot(days,nop1,nop2,nop3,baseline="wiggle")
plt.legend()
plt.show()
```

STEP PLOT MATPLOTLIB 

```python
x=["day1","day2","day3","day4","day5"]
y=[30,40,25,30,40]
plt.step(x,y,where="post"or"pre"or"mid",marker="o")
plt.show()
```

SUBPLOT MATPLOTLIB 

```python
x=[1,2,3,4,5]
y=[45,34,56,23,45]
plt.subplot(1,2,1)
plt.plot(x,y)
plt.show()
```
