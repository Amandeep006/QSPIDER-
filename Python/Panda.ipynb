MERGE, JOIN AND CONCATENATE IN PANDAS 

```python
data1={
    "EMPID":["E01","E02","E03","E04","E05"],
    "Name":["Ram","Shyam","Mohan","Sohan","Raju"],
    "Age":[23,42,12,24,42]
}

data2={
    "EMPID":["E01","E02","E03","E04","E05"],
    "SAL":[45000,56000,43000,23000,50000]
}

d1=pd.DataFrame(data1)
d2=pd.DataFrame(data2)
```

```python
#Table 1
d1
```

```python
# Table 2
d2
```

```python
# Now we are merging table 1 and 2 with the use of common table 
pd.merge(d1,d2,on="EMPID")
```

```python
# Now we are joining both the table using left and right join according to the need.
pd.merge(left=d1,right=d2,on="EMPID",how="left")
pd.merge(left=d1,right=d2,on="EMPID",how="right")
```

```python
# Now we concatenate both the table 
pd.concat([d1,d2])
```
