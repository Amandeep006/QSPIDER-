For each row, starting from the first, print numbers in descending order from n down to 1.
Each number in a row is repeated as many times as the current row index (starting from n).
Instead of printing each row on a new line, separate rows with -1.
Instead of a newline at the end of each row, print -1 to indicate row separation. After printing the entire pattern, end the output with -1.

For n= 3,
pattern:  3 3 3 2 2 2 1 1 1
                    3 3 2 2 1 1 
                    3 2 1

For n= 2,
pattern:  2 2 1 1
                    2 1

Input: 2

Output: [3, 3, 3, 2, 2, 2, 1, 1, 1, -1, 3, 3, 2, 2, 1, 1, -1, 3, 2, 1, -1]

45. Check Whether a Number is a Happy Number.  
   
Problem Statement:  
A number is called a Happy Number if repeatedly replacing the number by the sum of the squares of its digits
eventually produces 1. If the process enters a cycle without reaching 1, it is not happy

```python
n=int(input("Enter any number : "))
a=n
s=[]
f=0
while True :
    sum=0
    while n>0:
        rem=n%10
        sum=sum+rem**2
        n//=10
    if sum==1:
        f=1
        break
    elif sum in s:
        break
    else:
        s.append(sum)
        n=sum


if f==1:
    print(f"The given {a} number is Happy number.")
else:
    print(f"The given {a} number is not Happy number.")
```

```python
n1=int(input("Enter the numerator1 : "))
d1=int(input("Enter the denominator1 : "))
n2=int(input("Enter the numerator2 : "))
d2=int(input("Enter the denominator2 : "))

num=(n1*d2)+(n2*d1)
dem=d1*d2

if num>dem:
    small=dem
else:
    small=num

for i in range(1,small+1):
    if num%i==0 and dem%i==0:
        num=num//i
        dem=dem//i
print(f"Numerator : {num} and Denominator : {dem}")
```
