You are given a number n. You need to generate and print a pattern based on the given value of n.

For each row, starting from the first, print numbers in descending order from n down to 1.
Each number in a row is repeated as many times as the current row index (starting from n).
Instead of printing each row on a new line, separate rows with -1.
Instead of a newline at the end of each row, print -1 to indicate row separation. After printing the entire pattern, end the output with -1.

For n= 3,
pattern:  3 3 3 2 2 2 1 1 1
                    3 3 2 2 1 1 
                    3 2 1

For n= 2,
pattern:  2 2 1 1
                    2 1

Examples :

Input: 2
Output: [2, 2, 1, 1, -1, 2, 1, -1]
Input: 3
Output: [3, 3, 3, 2, 2, 2, 1, 1, 1, -1, 3, 3, 2, 2, 1, 1, -1, 3, 2, 1, -1]
Input: 1

45. Check Whether a Number is a Happy Number.  
   
Problem Statement:  
A number is called a Happy Number if repeatedly replacing the number by the sum of the squares of its digits
eventually produces 1. If the process enters a cycle without reaching 1, it is not happy

```python
n=int(input("Enter any number : "))
a=n
s=[]
f=0
while True :
    sum=0
    while n>0:
        rem=n%10
        sum=sum+rem**2
        n//=10
    if sum==1:
        f=1
        break
    elif sum in s:
        break
    else:
        s.append(sum)
        n=sum


if f==1:
    print(f"The given {a} number is Happy number.")
else:
    print(f"The given {a} number is not Happy number.")
```

```python
def fact(n):
    if n==0 or n==1:
        return 1
    return n*fact(n-1)


n=int(input("Enter the no. of people : "))
r=int(input("Enter the no. of selection : "))

nCr=fact(n)//(fact(r)*fact(n-r))

print(f"No. of ways : {nCr}")
```

50. Add Two Fractions and Simplify the Result.  
  
Problem Statement:  
Given the numerator and denominator of two fractions, calculate their sum and return the result in its simplest
form.

```python
n1=int(input("Enter the numerator1 : "))
d1=int(input("Enter the denominator1 : "))
n2=int(input("Enter the numerator2 : "))
d2=int(input("Enter the denominator2 : "))

num=(n1*d2)+(n2*d1)
dem=d1*d2

if num>dem:
    small=dem
else:
    small=num

for i in range(1,small+1):
    if num%i==0 and dem%i==0:
        num=num//i
        dem=dem//i
print(f"Numerator : {num} and Denominator : {dem}")
```

Write a function called highest_frequency that takes a list of integers as input and returns a list of integers that have the highest frequency. If multiple integers have the same highest frequency, all of them should be included in the list.

For example, given the list [1, 3, 2, 2, 1, 1, 4, 2, 4, 3], the function should return [1, 2], as both 1 and 2 appear 3 times in the list, which is the highest frequency among all integers.

```python
def highest_frequency(lst):
	frequency = {}
	max_freq = 0

	# Count the frequency of each element in the list
	for num in lst:
		frequency[num] = frequency.get(num, 0) + 1
		max_freq = max(max_freq, frequency[num])

	# Find all numbers with the highest frequency
	result = []
	for num, freq in frequency.items():
		if freq == max_freq:
			result.append(num)

	return result

print(highest_frequency([1,2,3,2,1]))
```

Question 40 — Check if a String is a Valid Identifier  
Problem Statement  
Given a string, determine whether it can be used as a valid Python identifier.  
Input  
A string S.  
Output  
Print Valid if S is a valid Python identifier; otherwise print Invalid.  
Examples  
Example 1:  
Input: student_name  
Output: Valid  
Example 2:  
Input: 123student  
Output: Invalid  

```python
s=input("Enter any string : ")
if s.isidentifier():
    print("valid")
else:
    print("Invalid")
```

```python
# WAP TO GET THE FOLLOWING OUTPUT 
# INPUT : "push maadi kushi padi"
# OUTPUT: {"push":"ph","maddi":"a","kushi":"s","padi":"pi"}


s=input("Enter any string : ")
p=s.split()
d={}
for i in p:
    if len(i)%2==0:
        o=i[0]+i[-1]
        d[i]=d.get(i,"")+o
    else:
        mid=len(i)//2
        d[i]=d.get(i,"")+i[mid]

print(d)
    
    
```

```python
# WAP TO GET THE FOLLOWING OUTPUT 
# S=["jiocinema.com","file.py","web.html","amazon.com","www.org"]
# out=["com","py","html","org"]

s=eval(input("Enter any string : "))
out=[]
for i in s:
    ex=i.split(".")[-1]
    if ex not in out:
        out.append(ex)
print(out)
```
