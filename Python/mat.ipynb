```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/kanchanchy/Data-Visualization-in-Python/refs/heads/main/data/stock_data.csv")
print(df)
```

1. PLOT BASIC LINE PLOT

```python

df.plot(x="Date",y="FB",label="facebook Stock Price", figsize=(15,10),linewidth=3)
plt.ylabel("Price[$]")
plt.title("My first plotting exercise")
plt.legend(loc="lower right")
plt.grid()
```

```python
df.plot(x="Date",y="NFLX",label="facebook Stock Price",figsize=(10,15),linewidth=3)
plt.ylabel("Price[$]")
plt.title("My first plotting exercise")
plt.legend(loc="lower right")
plt.grid()
```

2. PLOT SCATTERPLOT 

```python
# Read daily return data using pandas 
daily_df=pd.read_csv("https://raw.githubusercontent.com/kanchanchy/Data-Visualization-in-Python/refs/heads/main/data/stocks_daily_returns.csv")
print(daily_df)
```

```python
x=daily_df["FB"]
x
```

```python
y=daily_df["TWTR"]
y
```

```python
plt.figure(figsize=(15,19)) # figsize(width, height)
plt.scatter(x,y)
plt.grid()
```

3. PLOT PIE CHAR 

```python
value=[20,55,5,17,3]
color=['g','r','y','b','m']
label=["Apple","google","samsung","T","Amzn"]
explode=[0,0.2,0,0,0.2]

# use matplotlib to plot a pie chart
plt.figure(figsize=(10,10))
plt.pie(value,colors=color,labels=label,explode=explode)
plt.title("Stock Portfolio")
```

```python
values=[20,20,20,20,20]
colors=['g','r','y','b','m']
labels=["Apple","Google","T","TSLA","AMZN"]
explode=[0,0.2,0,0,0.2]

# use matplotlib to plot a pie chart 
plt.figure(figsize=(10,10))
plt.pie(values, colors=colors,labels=labels, explode=explode)
plt.title("Stock Portfolio")
```

4. PLOT HISTOGRAMS

```python
# A histogram represents data using bars of various heights.
# Each bar groups numbers into speccific ranges.
# Taller bars show that more data falls within that specific range.

mu=daily_df["FB"].mean()
sigma=daily_df["FB"].std()

num_bins=40
plt.figure(figsize=(15,9))
plt.hist(daily_df["FB"],num_bins,facecolor="blue"); # ; is to get rid of 
plt.grid()
plt.title(f"Histogram : mu= {mu}, sigma: {sigma}")
```

5. PLOT MULTIPLE PLOTS.

6. PLOT SUBPLOTS

```python
plt.figure(figsize=(20,10))
plt.subplot(1,2,1) # will have 1 row and 2 columns, we are plotting first one 
plt.plot(df["NFLX"],'r--') # r color, ---style
plt.grid()

plt.subplot(1,2,2) # will have 1 ros and 2 columns, we are plotting second one

plt.plot(df["FB"], "b.")
plt.grid()
```

```python
plt.figure(figsize=(20,10))
plt.subplot(2,1,1) # will have 2 rows and 1 column, we are plotting first one 
plt.plot(df["NFLX"],"r--") # r color, --style
plt.grid()

plt.subplot(2,1,2)
plt.plot(df["FB"],"b.")
plt.grid()
```
